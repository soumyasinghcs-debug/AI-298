```python
import heapq
# initial state
print("Enter Initial State:")
initial = []
for i in range(3):
    row = list(map(int, input().split()))
    initial = initial + row

# goal state
print("Enter Goal State:")
goal = []
for i in range(3):
    row = list(map(int, input().split()))
    goal = goal + row

initial = tuple(initial)
goal = tuple(goal)


# Manhattan Distance
def manhattan(state):
    distance = 0
    for i in range(9):
        if state[i] != 0:
            # Current position
            row = i // 3
            col = i % 3
            # Find position in goal
            goal_position = goal.index(state[i])
            goal_row = goal_position // 3
            goal_col = goal_position % 3

            distance = distance + abs(row - goal_row)
            distance = distance + abs(col - goal_col)

    return distance


# Find neighboring states
def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row = zero // 3
    col = zero % 3
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for move in moves:
        new_row = row + move[0]
        new_col = col + move[1]
        if new_row >= 0 and new_row < 3 and new_col >= 0 and new_col < 3:
            new_zero = new_row * 3 + new_col
            new_state = list(state)
            new_state[zero] = new_state[new_zero]
            new_state[new_zero] = 0
            neighbors.append(tuple(new_state))
    return neighbors

# A* Algorithm
def a_star(initial):
    pq = []
    g = 0
    h = manhattan(initial)
    f = g + h
    heapq.heappush(pq, (f, g, initial))
    parent = {}
    parent[initial] = None
    while pq:
        f, g, current = heapq.heappop(pq)
        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            path.reverse()
            return path
        for neighbor in get_neighbors(current):
            new_g = g + 1
            if neighbor not in parent:
                parent[neighbor] = current
                h = manhattan(neighbor)
                f = new_g + h
                heapq.heappush(pq, (f, new_g, neighbor))
    return None

solution = a_star(initial)

if solution is None:
    print("\nNo solution exists for this configuration!")
else:
    print("\nSolution:")
    i = 0
    for state in solution:
        print("\nStep", i)
        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])
        i = i + 1
    print("\nTotal moves:", len(solution) - 1)
```
