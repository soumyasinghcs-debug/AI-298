```python

def substitute(x, subs):
    while isinstance(x, str) and x in subs:
        x = subs[x]

    if isinstance(x, tuple):
        return tuple(substitute(i, subs) for i in x)

    return x


def occurs(var, x, subs):
    x = substitute(x, subs)

    if var == x:
        return True

    if isinstance(x, tuple):
        return any(occurs(var, i, subs) for i in x)

    return False


def unify(x, y, subs=None):
    if subs is None:
        subs = {}

    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if isinstance(x, str) and x.islower():
        if occurs(x, y, subs):
            return None
        subs[x] = y
        return subs

    if isinstance(y, str) and y.islower():
        if occurs(y, x, subs):
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y):
            return None

        for a, b in zip(x, y):
            subs = unify(a, b, subs)
            if subs is None:
                return None

        return subs

    return None

def parse(s):
    s = s.strip()

    if "(" not in s:
        return s

    name, args = s.split("(", 1)
    args = args.rstrip(")")
    return (name.strip(),) + tuple(
        parse(a.strip()) for a in args.split(",")
    )


# Main program
x = input("Enter first expression (X): ")
y = input("Enter second expression (Y): ")

x = parse(x)
y = parse(y)

result = unify(x, y)

if result is not None:
    print("\nUnification Successful!")
    print("Substitution:", result)
    print("Final X:", substitute(x, result))
    print("Final Y:", substitute(y, result))
else:
    print("\nUnification Failed!")
```
